```python
class Solution:
    def canConstruct(self, ransomNote, magazine):
        hash_map = {
            'a' : 0,
            'b' : 0,
            'c' : 0,
            'd' : 0,
            'e' : 0,
            'f' : 0,
            'g' : 0,
            'h' : 0,
            'i' : 0,
            'j' : 0,
            'k' : 0,
            'l' : 0,
            'm' : 0,
            'n' : 0,
            'o' : 0,
            'p' : 0,
            'q' : 0,
            'r' : 0,
            's' : 0,
            't' : 0,
            'u' : 0,
            'v' : 0,
            'w' : 0,
            'x' : 0,
            'y' : 0,
            'z' : 0
        }

        for ransom in ransomNote:
            hash_map[ransom] += 1


        for mag in magazine:
            hash_map[mag] -= 1

        if sum(hash_map.values()) == 0:
            return True
        else:
            return False
        
```

```python
s = Solution()
s.canConstruct(ransomNote = "aa", magazine = "aab")
```

```python
class Solution:
    def pivotArray(self, nums,pivot):
        left = []
        right = []
        mid = []

        for num in nums:
            if num < pivot:
                left.append(num)
            elif num > pivot:
                right.append(num)
            else:
                mid.append(num)

 

        
        return left  + mid + right
        
```

```python
s = Solution()
s.pivotArray(nums = [9,12,5,10,14,3,10], pivot = 10)
```
