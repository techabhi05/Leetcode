```python
class Solution:
    def canConstruct(self, ransomNote, magazine):
        hash_map = {
            'a' : 0,
            'b' : 0,
            'c' : 0,
            'd' : 0,
            'e' : 0,
            'f' : 0,
            'g' : 0,
            'h' : 0,
            'i' : 0,
            'j' : 0,
            'k' : 0,
            'l' : 0,
            'm' : 0,
            'n' : 0,
            'o' : 0,
            'p' : 0,
            'q' : 0,
            'r' : 0,
            's' : 0,
            't' : 0,
            'u' : 0,
            'v' : 0,
            'w' : 0,
            'x' : 0,
            'y' : 0,
            'z' : 0
        }

        for ransom in ransomNote:
            hash_map[ransom] += 1


        for mag in magazine:
            hash_map[mag] -= 1

        if sum(hash_map.values()) == 0:
            return True
        else:
            return False
        
```

```python
s = Solution()
s.canConstruct(ransomNote = "aa", magazine = "aab")
```

```python
class Solution:
    def removeDuplicates(self, s , k):
        pocket = []

        for ch in s:
            if len(pocket) == 0 or pocket[-1] != ch:
                pocket.append([ch,1])
            else:
                if pocket[-1][1] == k-2:
                    pocket.pop()
                else:
                    pocket[-1][1] += 1
        return pocket
                    
```

```python
s = Solution()
s.removeDuplicates(s = "deeedbbcccbdaa", k = 3)
```
